- To generate a synthetic marketing data generation engine that simulates real-world omnichannel marketing campaigns, customer engagement, and revenue outcomes with realistic business constraints.

**Import and Configuration**

```python
import numpy as np
import random

random.seed(1)
np.random.seed(1)
```

```python
brands=['Nike','Adidas','Puma','Reebok','Under Armour']
categories=['Shoes','Apparel','Accesories']

channels=['Facebook','Instagram','Google','YouTube','Email','Display','TV','Radio']
channel_types={"Google": "Search",
                "Facebook": "Social",
                "Instagram": "Social",
                "YouTube": "Video",
                "Email": "Owned",
                "Display": "Display",
                "TV": "Traditional",
                "Radio": "Traditional"}

audiences=['Students','Young Professionals','Parents','Fitness Enthusiasts','Gamers','Corporate']
audience_probability = [0.15, 0.30,0.20,0.20,0.10,0.05]

cities = ["Delhi","Mumbai","Bangalore","Hyderabad","Chennai","Pune","Kolkata"]
city_state = {"Delhi":"Delhi",
              "Mumbai":"Maharashtra",
              "Pune":"Maharashtra",
              "Bangalore":"Karnataka",
              "Hyderabad":"Telangana",
              "Chennai":"Tamil Nadu",
              "Kolkata":"West Bengal"}

campaign_types=['Brand Awareness','Lead Generation','Product Launch','Remarketing','Seasonal','Flash Sale']

project_start = datetime(2024,1,1)
project_end = datetime(2025,12,31)

print("Configuration Loaded Successfully")
```

**Channel Table**

```python
import pandas as pd 

channel_rows = []
for i, channel in enumerate(channels, start=1):
    channel_rows.append({"channel_id": f"CH{i:03}",
                        "channel_name": channel,
                        "channel_type": channel_types[channel]})
channels_df = pd.DataFrame(channel_rows)
channels_df
```

```python
channels_df.to_csv("data/raw/channels.csv",index=False)
```

- **campaign table**

```python
from datetime import datetime,timedelta

num_campaigns=100
budget_ranges = {"Brand Awareness": (300000, 700000),
                 "Lead Generation": (150000, 400000),
                 "Product Launch": (500000, 900000),
                 "Remarketing": (100000, 250000),
                 "Seasonal": (250000, 600000),
                 "Flash Sale": (80000, 180000)}
duration_choices = [30, 45, 60, 90] #days
campaign_status = ["Completed", "Active"]
status_probability = [0.9, 0.1] #90% completed 10% Active
```

```python
campaign_rows = []
for i in range(1, num_campaigns + 1):
    campaign_id = f"CA{i:03}"
    campaign_type = random.choice(campaign_types)
    
    brand = random.choice(brands)
    category = random.choice(categories)
    campaign_name = f"{brand} {campaign_type}"

    
    budget_min, budget_max = budget_ranges[campaign_type]
    campaign_budget = random.randint(budget_min, budget_max)
    
    
    channel_id = f"CH{random.randint(1, len(channels)):03}"
    
    random_days = random.randint(0,(project_end - project_start).days)
    campaign_start_date = project_start + timedelta(days=random_days)
    duration = random.choice(duration_choices)
    campaign_end_date = campaign_start_date + timedelta(days=duration)

    if campaign_end_date > project_end:
        campaign_end_date = project_end
        
    status = random.choices(campaign_status,weights=status_probability,k=1)[0]

    campaign_rows.append({"campaign_id": campaign_id,
                          "campaign_name": campaign_name,
                          "campaign_type": campaign_type,
                          "brand": brand,
                          "product_category": category,
                          "channel_id": channel_id,
                          "campaign_start_date": campaign_start_date.strftime("%Y-%m-%d"),
                          "campaign_end_date": campaign_end_date.strftime("%Y-%m-%d"),
                          "campaign_budget": campaign_budget,
                          "campaign_status": status})
campaign_df = pd.DataFrame(campaign_rows)
campaign_df.head()
```

- **external_facters Table**

```python
def get_weather(month):
    if month in [6, 7, 8, 9]:
        return random.choices(["Rainy", "Cloudy", "Sunny"],weights=[0.60, 0.25, 0.15],k=1)[0]
    elif month in [10, 11, 12, 1, 2]:
        return random.choices(["Sunny", "Cloudy", "Rainy"],weights=[0.60, 0.25, 0.15],k=1)[0]
    else:
        return random.choice(["Sunny", "Cloudy", "Rainy"])
```

```python
date_range = pd.date_range(start=project_start,end=project_end,freq="D")

external_rows = []
decline_start = datetime(2025, 7, 1)

for i, current_date in enumerate(date_range, start=1):
    factor_id = f"EF{i:03}"
    month = current_date.month
    holiday = random.choices(["Yes", "No"],weights=[0.08, 0.92],k=1)[0]
    
    if month in [10, 11, 12]:
        festival = random.choices(["Yes", "No"],weights=[0.25, 0.75],k=1)[0]
    else:
        festival = random.choices(["Yes", "No"],weights=[0.04, 0.96],k=1)[0]    
        
    weather = get_weather(month)

    if current_date >= decline_start: #Competitor spend increases from the point of sales decline
        competitor_spend = random.randint(450000,800000)
    else:
        competitor_spend = random.randint(150000,500000)

    if festival == "Yes":
        discount = random.choice([20, 25, 30, 35, 40])
    else:
        discount = random.choice([0, 5, 10, 15, 20])

    external_rows.append({"factor_id": factor_id,
                          "date": current_date.strftime("%Y-%m-%d"),
                          "holiday_flag": holiday,
                          "festival_flag": festival,
                          "weather": weather,
                          "competitor_spend": competitor_spend,
                          "discount_percentage": discount})
    
external_df = pd.DataFrame(external_rows)
external_df.head()
    
```

- **marketing_performace Table**

Defining campaign_type behaviours that resembles real-world business

```python
num_records=500000

campaign_rules = {
     #to provide maximum visibility and reach, with low values
    "Brand Awareness": {"ctr": (0.01,0.025),
                        "cvr": (0.01,0.025),
                        "aov": (2500,4500)},

    #to provide maximum reaches
    "Lead Generation": {"ctr": (0.03,0.05), 
                        "cvr": (0.03,0.06),
                        "aov": (3000,6000)},

    "Product Launch": {"ctr": (0.02,0.04),
                       "cvr": (0.02,0.05),
                       "aov": (3500,7000)},

    #to show ads only to the reaches, thus ctr and cvr will be higher
    "Remarketing": {"ctr": (0.05,0.08),
                    "cvr": (0.06,0.10),
                    "aov": (4000,8000)},
    
    "Seasonal": {"ctr": (0.03,0.06),
                 "cvr": (0.04,0.08),
                 "aov": (2500,5500)},

    #to readvertise and will have higher aov
    "Flash Sale": {"ctr": (0.05,0.09),
                   "cvr": (0.05,0.09),
                   "aov": (1800,4000)}}
```

```python
campaign_lookup = campaign_df.set_index("campaign_id").to_dict("index")
customer_lookup = customers_df.set_index("customer_id").to_dict("index")
external_lookup = external_df.set_index("date").to_dict("index")

campaign_ids = campaign_df["campaign_id"].tolist()
customer_ids = customers_df["customer_id"].tolist()

decline_start = datetime(2025,7,1)

marketing_rows = []
performance_counter=1
while len(marketing_rows) < num_records:
    
    campaign_id = random.choice(campaign_ids)
    campaign = campaign_lookup[campaign_id]
    
    customer_id = random.choice(customer_ids)
    customer = customer_lookup[customer_id]

    campaign_start = datetime.strptime(campaign["campaign_start_date"],"%Y-%m-%d")
    campaign_end = datetime.strptime(campaign["campaign_end_date"],"%Y-%m-%d")

    interaction_date = campaign_start + timedelta(days=random.randint(0,(campaign_end-campaign_start).days))
    interaction_date_str = interaction_date.strftime("%Y-%m-%d")

    # Simulate reduced customer engagement after July 2025
    if interaction_date >= decline_start:
        if random.random() < 0.10:
            continue
    performance_id = f"MP{performance_counter:08}"
    performance_counter += 1
    
    external = external_lookup[interaction_date_str]
    
    factor_id = external["factor_id"]
    
    competitor_spend = external["competitor_spend"]
    
    if interaction_date >= decline_start:
        competitor_spend = int(competitor_spend * 1.25)
        
    discount = external["discount_percentage"]
    
    weather = external["weather"]
    
    festival = external["festival_flag"]
    
    holiday = external["holiday_flag"]

    rules = campaign_rules[campaign["campaign_type"]]
    min_ctr,max_ctr = rules["ctr"]
    min_cvr,max_cvr = rules["cvr"]
    min_aov,max_aov = rules["aov"]

    campaign_duration = (campaign_end - campaign_start).days + 1
    
    daily_budget = campaign["campaign_budget"] / campaign_duration
    spend = random.randint(int(daily_budget*0.70),int(daily_budget*1.10))

    if festival == "Yes":
        spend *= 1.15
    if competitor_spend > 600000:
        spend *= 0.95
    if interaction_date >= decline_start:
        spend *= 0.90      # 10% lower media budget
    spend=int(spend)
    
    cpm = random.randint(150,400)
    if interaction_date >= decline_start:
        cpm = int(cpm * 1.10)

    impressions = int((spend/cpm)*1000)

    reach = random.randint(int(impressions*0.60),impressions)
    if interaction_date >= decline_start:
        reach = int(reach * 0.90)

    frequency = round(impressions/max(reach,1),2)

    ctr = random.uniform(min_ctr, max_ctr)
    if weather == "Rainy":
        ctr *= 0.95
    elif weather == "Cloudy":
        ctr *= 0.98

    if discount >= 30:
        ctr *= 1.10
    elif discount >= 20:
        ctr *= 1.05

    if interaction_date >= decline_start:
        ctr *= 0.88

    clicks = max(1,int(impressions * ctr))

    cvr = random.uniform(min_cvr,max_cvr)
    if holiday == "Yes":
        cvr *= 1.05
        
    if discount >= 30:
        cvr *= 1.15
    elif discount >= 20:
        cvr *= 1.08

    if competitor_spend > 600000:
        cvr *= 0.95
    if interaction_date >= decline_start:
        cvr *= 0.88

    conversions = max(1,int(clicks * cvr))

    aov = random.randint(min_aov,max_aov)
    if festival == "Yes":
        aov *= 1.08
    if interaction_date >= decline_start:
        aov *= 0.95
    aov=int(aov)

    revenue = int(conversions * aov)

    marketing_rows.append({"performance_id": performance_id,
                           "interaction_date": interaction_date.strftime("%Y-%m-%d"),
                           "campaign_id": campaign_id,
                           "customer_id": customer_id,
                           "factor_id": factor_id,
                           "spend": spend,
                           "impressions": impressions,
                           "reach": reach,
                           "frequency": frequency,
                           "clicks": clicks,
                           "conversions": conversions,
                           "revenue": revenue})

marketing_df = pd.DataFrame(marketing_rows)
marketing_df.head()
```
